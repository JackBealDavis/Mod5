```python
def diff(df_use):
    df_clean = df_use.copy()
    df_clean['diff_days'] = (df_use['Book Returned'] - df_use['Book checkout']).dt.days
    return df_clean
```

```python
import pandas as pd

df = pd.read_csv("BooksClean.csv")

df['Book checkout'] = pd.to_datetime(df['Book checkout'], errors='coerce')
df['Book Returned'] = pd.to_datetime(df['Book Returned'], errors='coerce')


df_clean = diff(df_use=df)

df_clean.to_csv("BooksClean.csv")

df_clean.info()
df_clean.head()
```
